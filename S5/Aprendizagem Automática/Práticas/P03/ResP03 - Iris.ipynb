# Código Original

```python
import numpy as np

from sklearn.datasets import load_iris

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix

import matplotlib.pyplot as plt
```

```python
#load data
iris = load_iris()
```

```python
#check data
print("iris.keys(): \n{}".format(iris.keys()))
print("Shape of iris data: {}".format(iris.data.shape))
print("Sample counts per class:\n{}".format(
	{n: v for n, v in zip(iris.target_names, np.bincount(iris.target))}))
print("Feature names:\n{}".format(iris.feature_names))
```

```python
#split data
X_train, X_test, y_train, y_test = train_test_split(
	iris['data'], iris['target'], random_state=0)
```

```python
#build model
knn = KNeighborsClassifier()
knn.fit(X_train, y_train)
```

```python
#use model
pred_knn = knn.predict(X_test)
```

```python
#matrix confusion
confusion = confusion_matrix(y_test, pred_knn)
print("Confusion matrix:\n{}".format(confusion))
print("Test set accuracy: {:.2f}".format(knn.score(X_test, y_test)))
```

## 6.

Utilizando a biblioteca "matplotlib" (https://matplotlib.org/stable/) faça um gráfico do desempenho (exatidão) em função do nº de exemplos do conjunto de treino (30, 50, 70, 90, completo).

```python
treinos = [30, 50, 70, 90, 112]
exatidao = []

for size in treinos:
    knn = KNeighborsClassifier()
    knn.fit(X_train[:size], y_train[:size])
    exatidao.append(knn.score(X_test, y_test))

plt.plot(treinos, exatidao)
plt.xlabel("Nº de exemplos de treino")
plt.ylabel("Exatidão")
plt.title("Desempenho do KNN vs Tamanho do Treino")
plt.show()
```


## 7.

Repita o exercício modificando o nº de vizinhos e observe o resultado.

```python
treinos = [30, 50, 70, 90, 112]
neighbors = [1, 2, 3, 5, 7]

plt.figure()

for n in neighbors:
    exatidao = []

    for size in treinos:
        knn = KNeighborsClassifier(n_neighbors=n)
        knn.fit(X_train[:size], y_train[:size])
        exatidao.append(knn.score(X_test, y_test))
    plt.plot(treinos, exatidao, marker='o', label=f"{n} vizinhos")


plt.xlabel("Nº de Exemplos de Treino")
plt.ylabel("Exatidão")
plt.title("Desempenho do KNN vs Tamanho do Treino vs Nº de Vizinhos")
plt.legend()
plt.show()
```
