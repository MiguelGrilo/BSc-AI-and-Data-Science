# Exercícios

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, classification_report
```

## 1.

Carregue o conjunto de dados "wine" usando a função "load_wine()".

```python
#load data
wine = load_wine()
```

## 2.

Faça uma divisão treino/teste com os valores por omissão e confirme o número de exemplos e
atributos de cada conjunto.

```python
#split data
X_train, X_test, y_train, y_test = train_test_split(
	wine.data, wine.target, stratify=wine.target, random_state=0)
```

```python
#número de exemplos e atributos de cada conjunto
print("Conjunto de treino:", X_train.shape)
print("Conjunto de teste:", X_test.shape)
```

## 3.

Construa um classificador usando um algoritmo à sua escolha. Obtenha a matriz de confusão.
Quantos exemplos tem cada uma das classes? Quantos exemplos foram mal classificados?

```python
classificador = DecisionTreeClassifier(random_state=0)
classificador.fit(X_train, y_train)

pred = classificador.predict(X_test)
matriz_confusao = confusion_matrix(y_test, pred)

print("Matriz de Confusão:\n", matriz_confusao)

#real -> linhas
#pred -> colunas
```

**Classe 1:** 15 exemplos (3 exemplo mal classificado)

**Classe 2:** 18 exemplos (1 exemplo mal classificado)

**Classe 3:** 12 exemplos (1 exemplo mal classificado)

**Total de Exemplos:** 45 exemplos (5 exemplos mal classificados)

## 5.

Calcule a exatidão, precisão, cobertura, e F1, considerando a micro-média, macro-média, e média
ponderada.

```python
exatidao = accuracy_score(y_test, pred)

prec_macro = precision_score(y_test, pred, average='macro')
cob_macro = recall_score(y_test, pred, average='macro')
f1_macro = f1_score(y_test, pred, average='macro')

prec_micro = precision_score(y_test, pred, average='micro')
cob_micro = recall_score(y_test, pred, average='micro')
f1_micro = f1_score(y_test, pred, average='micro')

prec_ponderada = precision_score(y_test, pred, average='weighted')
cob_ponderada = recall_score(y_test, pred, average='weighted')
f1_ponderada = f1_score(y_test, pred, average='weighted')

df = pd.DataFrame({
    "Métrica": ["Macro", "Micro", "Weighted"],
    "Exatidão": [exatidao, exatidao, exatidao],
    "Precisão": [prec_macro, prec_micro, prec_ponderada],
    "Cobertura": [cob_macro, cob_micro, cob_ponderada],
    "F1": [f1_macro, f1_micro, f1_ponderada]
})

print(df.to_string(index=False))
```

## 6.

Observe o funcionamento da função sklearn.metrics.classification_report

```python
print(classification_report(y_test, pred, target_names=wine.target_names))
```
